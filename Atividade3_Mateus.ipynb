```python
#4. Lista de números: pares e ímpares INSTRUÇÃO PEDAG ÓG ICA PA
numeros = []
pares = []
impares = []

print("Digite 10 números:")
for i in range(10):
  num = int(input(f"Digite o {i + 1}° número:"))
  numeros.append(num)

for n in numeros:
  if n % 2 == 0:
     pares.append(n)
  else:
     impares.append(n)

print("Números digitados: ", numeros)
print("Números pares: ", pares)
print("Números ímpares: ", impares)
```

```python
#6. Nomes sem repetição
nomes = []

while True:
    nome = input("Digite um nome (ou FIM para encerrar): ")

    if nome.upper() == "FIM":
        break

    if nome not in nomes:
        nomes.append(nome)

print("\nLista de nomes cadastrados:")
print(nomes)
print("Quantidade de nomes diferentes:", len(nomes))
```

```python
numeros = []

print("Digite 6 números inteiros:")
for i in range(6):
    num = int(input(f"Digite o {i+1}º número: "))
    numeros.append(num)

tupla = tuple(numeros)

qtd_cinco = tupla.count(5)

if 5 in tupla:
    posicao = tupla.index(5)
else:
    posicao = "não encontrado"

qtd_pares = 0
for num in tupla:
    if num % 2 == 0:
        qtd_pares += 1

print("\nTupla formada:", tupla)
print("Quantidade de vezes que o 5 apareceu:", qtd_cinco)
print("Posição da primeira ocorrência do 5:", posicao)
print("Quantidade de valores pares:", qtd_pares)
```

```python
contagem = {}

print("Digite 8 palavras:")
for i in range(8):
    palavra = input(f"Palavra {i+1}: ").lower()

    if palavra in contagem:
        contagem[palavra] = contagem[palavra] + 1
    else:
        contagem[palavra] = 1

print("\nContagem das palavras:")
for palavra, quantidade in contagem.items():
    print(f"{palavra}: {quantidade}")
```

```python
estoque = {
    "arroz": 20,
    "feijao": 15,
    "macarrao": 30,
    "oleo": 10
}

print("Estoque atual:", estoque)

produto = input("Digite o nome do produto: ").lower()
quantidade = int(input("Digite a quantidade vendida: "))

if produto in estoque:
    if estoque[produto] >= quantidade:
        estoque[produto] = estoque[produto] - quantidade
        print("Venda realizada com sucesso!")
    else:
        print("Estoque insuficiente!")
else:
    print("Produto não cadastrado!")

print("\nEstoque final:", estoque)
```

```python
turma = []

for i in range(5):
    print(f"\nAluno {i+1}:")
    nome = input("Nome: ")
    nota1 = float(input("Nota 1: "))
    nota2 = float(input("Nota 2: "))

    media = (nota1 + nota2) / 2

    aluno = {
        "nome": nome,
        "nota1": nota1,
        "nota2": nota2,
        "media": media
    }

    turma.append(aluno)

print("\n--- Resultado da turma ---")
for aluno in turma:
    print(f"Nome: {aluno['nome']} | Média: {aluno['media']:.2f}")
```

```python
produtos = []

print("Cadastro de 5 produtos:")
for i in range(5):
    print(f"\nProduto {i+1}:")
    codigo = int(input("Código: "))
    nome = input("Nome: ")
    preco = float(input("Preço: "))

    produto = (codigo, nome, preco)
    produtos.append(produto)

print("\nProdutos cadastrados:", produtos)

codigo_busca = int(input("\nDigite o código do produto que deseja buscar: "))

encontrado = False
for produto in produtos:
    if produto[0] == codigo_busca:
        print(f"Produto encontrado: {produto[1]} - R$ {produto[2]:.2f}")
        encontrado = True
        break

if not encontrado:
    print("Código não encontrado.")
```

```python
estoque = []

print("Cadastro de 5 itens de estoque:")
for i in range(5):
    print(f"\nItem {i+1}:")
    codigo = int(input("Código: "))
    nome = input("Nome: ")
    quantidade = int(input("Quantidade: "))
    preco = float(input("Preço unitário: "))

    item = {
        "codigo": codigo,
        "nome": nome,
        "quantidade": quantidade,
        "preco": preco
    }
    estoque.append(item)

# (a) Mostrar todos os itens
print("\n--- Itens cadastrados ---")
for item in estoque:
    print(f"Código: {item['codigo']} | Nome: {item['nome']} | "
          f"Qtd: {item['quantidade']} | Preço: R$ {item['preco']:.2f}")

# (b) Calcular o valor total do estoque
valor_total = 0
for item in estoque:
    valor_total += item["quantidade"] * item["preco"]

print(f"\nValor total do estoque: R$ {valor_total:.2f}")

# (c) Identificar o item de maior valor (sem max())
maior_valor = -1
item_maior = None

for item in estoque:
    valor_item = item["quantidade"] * item["preco"]
    if valor_item > maior_valor:
        maior_valor = valor_item
        item_maior = item

print(f"\nItem de maior valor em estoque:")
print(f"Nome: {item_maior['nome']} | Valor: R$ {maior_valor:.2f}")

# (d) Criar a tupla resumo
resumo = (len(estoque), valor_total)
print(f"\nTupla resumo: {resumo}")
```
