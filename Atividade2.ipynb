```python
#4. Estacionamento
horas = int(input("Digite a quantidade de horas de estacionamento:"))

if horas <= 2:
    valor = 8.00
elif horas <= 5:
    valor = 8.00 + 3.00 * (horas - 2)
else:
    valor = 17.00 + 2.00 * (horas - 5)

print(f"R$ {valor:.2f}")
```

```python
#5. Máquina de venda de ingresso
preco = 25
total = 0

while total < preco:
    valor = int(input("Digite o valor: "))
    total = total + valor

troco = total - preco

print("Total inserido:", total)
print("Troco:", troco)
```

```python
#6. Maior valor e sua posição
maior = -999999   # valor bem baixo para começar
posicao = 0

for i in range(1, 9):          # de 1 até 8
    num = int(input(f"Digite o {i}º número: "))

    if num > maior:
        maior = num
        posicao = i

print("Maior valor:", maior)
print("Primeira posição:", posicao)
```

```python
#7. Segundo maior valor
maior = -999999
segundo = -999999

for i in range(1, 9):
    num = int(input(f"Digite o {i}º número: "))

    if num > maior:
        segundo = maior      # o antigo maior vira segundo
        maior = num          # atualiza o maior
    elif num > segundo:
        segundo = num        # atualiza só o segundo

print("Maior:", maior)
print("Segundo maior:", segundo)
```

```python
#8. Detectando repetição consecutiva 7
Pythontem_repeticao = False
anterior = int(input("Digite o 1º número: "))

for i in range(2, 9):
    atual = int(input(f"Digite o {i}º número: "))

    if atual == anterior:
        tem_repeticao = True

    anterior = atual

if tem_repeticao:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")
```

```python
#9. Controle de saldo
saldo = 500.00

for i in range(1, 7):
    print(f"\nOperação {i}")
    tipo = int(input("Tipo (1=Depósito / 2=Saque): "))
    valor = float(input("Valor: "))

    if tipo == 1:
        saldo = saldo + valor
    elif tipo == 2:
        if valor <= saldo:
            saldo = saldo - valor
        else:
            print("Saldo insuficiente")

print("\nSaldo final: R$", saldo)
```

```python
#10. Caixa eletrônico simplificado
valor = int(input("Digite o valor: "))

cedulas = [100, 50, 20, 10, 5, 2, 1]

print(f"\nValor: R$ {valor}")
print("Cédulas necessárias:")

for cedula in cedulas:
    quantidade = valor // cedula
    if quantidade > 0:
        print(f"{quantidade} de R$ {cedula}")
    valor = valor % cedula
```

```python
#11. Número primo
n = int(input("Digite um número maior que 1: "))

eh_primo = True

for i in range(2, n):
    if n % i == 0:
        eh_primo = False
        break

if eh_primo:
    print("É primo")
else:
    print("Não é primo")
```

```python
#12. Divisores de um número
n = int(input("Digite um número positivo: "))

quantidade = 0

print("Divisores:", end=" ")

for i in range(1, n + 1):
    if n % i == 0:
        print(i, end=" ")
        quantidade = quantidade + 1

print("\nQuantidade de divisores:", quantidade)
```

```python
#13. Números primos em um intervalo
a = int(input("Digite A: "))
b = int(input("Digite B: "))

quantidade = 0

print("Números primos:", end=" ")

for num in range(a, b + 1):
    if num > 1:
        eh_primo = True
        for i in range(2, num):
            if num % i == 0:
                eh_primo = False
                break
        if eh_primo:
            print(num, end=" ")
            quantidade = quantidade + 1

print("\nQuantidade de primos encontrados:", quantidade)
```

```python
#14. Jogo com limite de tentativas
secreto = 37
tentativas = 0
acertou = False

while tentativas < 5 and not acertou:
    palpite = int(input("Digite seu palpite: "))
    tentativas = tentativas + 1

    if palpite == secreto:
        print("Parabéns! Você acertou!")
        acertou = True
    elif palpite < secreto:
        print("O número secreto é maior.")
    else:
        print("O número secreto é menor.")

if not acertou:
    print("Você usou todas as 5 tentativas. O número era 37.")
```

```python
#15. Desafio final - caixa de atendimento
saldo = 200.00
qtd_entradas = 0
qtd_retiradas = 0
qtd_recusadas = 0
total_entrou = 0.0
total_saiu = 0.0

codigo = int(input("Digite o código (1=entrada, 2=retirada, 0=sair): "))

while codigo != 0:
    if codigo == 1:
        valor = float(input("Valor da entrada: "))
        saldo = saldo + valor
        qtd_entradas = qtd_entradas + 1
        total_entrou = total_entrou + valor

    elif codigo == 2:
        valor = float(input("Valor da retirada: "))
        if valor <= saldo:
            saldo = saldo - valor
            qtd_retiradas = qtd_retiradas + 1
            total_saiu = total_saiu + valor
        else:
            print("Saldo insuficiente")
            qtd_recusadas = qtd_recusadas + 1

    codigo = int(input("Digite o código (1=entrada, 2=retirada, 0=sair): "))

print("\n----- Relatório Final -----")
print("Saldo final: R$", saldo)
print("Quantidade de entradas:", qtd_entradas)
print("Quantidade de retiradas realizadas:", qtd_retiradas)
print("Quantidade de retiradas recusadas:", qtd_recusadas)
print("Total que entrou: R$", total_entrou)
print("Total que saiu: R$", total_saiu)
```
